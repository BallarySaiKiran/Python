```python
'''append
-->it take one value as argument
-->it is used to add the elments at the end of the list
'''
```

```python
x=[10,20,30,40]
x.append(50)
print(x)
```

```python
'''extend
-->it take one argument(list,tuple,set)
-->it is used to add the multiple elments at the end of the list
'''
```

```python
x=[21,65,65,97]
x.extend([1,2,3,4])
print(x)
```

```python
'''insert
-->it used to insert elments in a list before the index postition
-->it expect index position and element'''
```

```python
x=[65,'kiran',0.5]
x.insert(0,True)
print(x)
```

```python
'''pop
-->it is used to delete the element based on index position
-->it will delete last elemet if we are not passing any index value'''
```

```python
x=['sai',23,63,00.9]
x.pop()
print(x)  # last element is deleted
```

```python
x.pop(0)
print(x)
```

```python
'''remove
-->it is used to delete the element based on list element
'''
```

```python
x=[75,'fgtds',True,257]
x.remove(75)
print(x)
```

```python
'''index
-->it find the index value of the element
'''
```

```python
'''count
-->it is used to count how many times the value is appears
'''
```

```python
x=[8,3,5,6,8,2,5,1]
print(x.count(5)) # 5 is repeted 2 times
```

```python
'''reverse
-->it is used to reverse the entire list
'''
```

```python
x=['rajesh','shiva',45,True,0.664]
x.reverse()
print(x)
```

```python
'''sort
-->it is used to arrage all the elements in ascending or descending order
--> by default it is ascending order
'''
```

```python
x=[1,4,87,34,7] 
x.sort()
print(x)
```

```python
x=[1,4,87,34,7] 
x.sort(reverse=True)
print(x)  # descending order
```

```python
'''copy
-->it is used to copy all the elements from the list
'''
```

```python
'''clear
-->it is used to remove all elements
'''
```

```python
x=[123,54,5,876]
y=x.clear()
print(y)
```
