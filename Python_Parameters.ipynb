```python
'''1.position parameter
-->number of parameters = number of arguments.
-->if we interchnage the arguments result will change.'''
```

```python
def sum1(x,y):  
    print(x+y)
sum1(2,5)
```

```python
def sub1(x,y):  
    print(x-y)
sub1(2,5)
```

```python
def sub1(x,y):  # output is changed
    print(x-y)
sub1(5,2)
```

```python
'''2.Keyword Parameter
-->number of parameters = number of arguments.
-->if we interchnage the arguments result will not change.'''
def sum1(x,y):
    print(x+y)
sum1(x=10,y=20)
```

```python
def sum1(x,y):
    print(x+y)
sum1(y=20,x=10)
```

```python
def sub1(x,y):
    print(x-y)
sub1(y=20,x=10)
```

```python
def sub1(x,y):  # output is not changed
    print(x-y)
sub1(x=10,y=20)
```

```python
'''creating a function with one keyword parameter and one position parameter
-->name is keyword parameter and age is position parameter
-->first we have to give position argument '''
def bio_data(age,name):
    print(name)
    print(age)
bio_data(23,name='Sai Kiran')
```

```python
def bio_data(name,age):
    print(name)
    print(age)
bio_data(23,name='Sai Kiran')
# note:position arguments will not follow keyword arguments
```

```python
'''3.Default Parameter
-->it is also known as optional parameter
-->it contain default value.if user is not passing any argument default value will be stored inside 
   the default parameter'''
```

```python
'''creating a function with 3 parameters name,age,pan.no
here:pan.no is default parameter '''

def data(name,age,pan_no='not available'):
    print(name)
    print(age)
    print(pan_no)
data('ravi',25,'jxrpk5545p')
```

```python
def data(name,age,pan_no='not available'):
    print(name)
    print(age)
    print(pan_no)
data('ravi',25)  #user is not passing pan_no it taken default parameter
```

```python
'''4.Arbitary Parameter
-->sometimes we do not know in advance how many number of arguments that will be passed into a 
   parameter of a function in this case Arbitary parameter is used
-->we use * before parameter name to denote Arbitary parameter'''
```

```python
def m1(name,email,*hobbies):
    print(name)
    print(email)
    print(hobbies)
m1('sai','sai@gmail.com','cooing','playig games')
```

```python
def bio_data(name,pan_no='not available',*hobbies,age):
    print(name)
    print(age)
    print(hobbies)
    print(pan_no)
bio_data('Sai Kiran','hg543dghd','cooking','dancing',age=25)
```

```python
'''4.Keyward arbitary Parameter
-->Keyward arbitary parameter will accept the multiple pair elements
-->we use ** before parameter name to denote keyward arbitary parameter'''
```

```python
def m2(name,**address):
    print(name)
    print(address)
m2('vignesh',house_no=1/662,street='bommanahalli',city='Banglore')
```
